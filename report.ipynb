# Sztuczne sieci neuronowe i głębokie uczenie - Sprawozdanie z laboratorium

## Temat:

Konwolucyjne sieci neuronowe w zagadnieniu
klasyfikacji obrazów I

### Cel ćwiczenia:

Celem ćwiczenia było zbudowanie i przetestowanie klasyfikatora obrazów kwiatów z użyciem transfer learningu (MobileNetV2), porównanie wpływu treningu w dwóch fazach (zamrożona baza + fine-tuning) oraz ocena jakości modelu na podstawie wykresów metryk i macierzy pomyłek.

- TensorFlow Datasets 4.9.10
- NumPy 2.3.5
- Matplotlib 3.10.6
- Seaborn 0.13.2
- scikit-learn n/d
- Jupyter Notebook

### Wnioski

1. Zastosowanie transfer learningu (MobileNetV2) umożliwiło skuteczne uczenie klasyfikatora bez trenowania sieci od zera.
2. Podział treningu na fazę z zamrożoną bazą i fazę fine-tuningu stabilizuje uczenie i zwykle poprawia metryki walidacyjne.
3. Analiza wykresów accuracy/loss pozwala ocenić, czy pojawia się przeuczenie oraz czy fine-tuning przyniósł poprawę.
4. Confusion Matrix ujawnia pary klas o podobnych cechach wizualnych, które model najczęściej myli.
